````python
from mlx_vlm import load, generate
from mlx_vlm.utils import load_image, load_config
from mlx_vlm.prompt_utils import apply_chat_template
import json
````

````python
# 1. Define the model path (This will auto-download from Hugging Face on first run)
model_path = "mlx-community/Qwen2-VL-2B-Instruct-4bit"
````

````python
# 2. Load the model and its processor onto your Mac's unified memory
model, processor = load(model_path)
config = load_config(model_path)
````

````python
image = ["/Users/akhileshmadyastha/Desktop/projects/document-understanding/gold/images/1_content_p001.png"]
````

````python
prompt = (
    """ Extract every field from this marriage certificate. 
     Return the result in this format : 

    {
  "source_pdf": "1_content.pdf",
  "source_page": 1,
  "jurisdiction_note": null,
  "state": null,
  "city": null,
  "license_number": null,
  "application_number": null,
  "marriage_day": null,
  "marriage_month": null,
  "marriage_year": null,
  "marriage_year_text": null,
  "marriage_location": null,
  "groom": {
    "full_name": null,
    "residence": null,
    "age": null,
    "color": null,
    "nativity": null,
    "occupation": null,
    "marital_status": null
  },
  "bride": {
    "full_name": null,
    "residence": null,
    "age": null,
    "color": null,
    "nativity": null,
    "occupation": null,
    "marital_status": null
  },
  "groom_consent_by": null,
  "bride_consent_by": null,
  "parents_consent_filed": null,
  "relationship_between_parties": null,
  "officiant": {
    "signature": null,
    "official_character": null,
    "address": null
  }
}
    
    """
)
````

````python
formatted_prompt = apply_chat_template(
    processor,
    config,
    prompt,
    num_images=1,
)
````

````python
output = generate(
    model,
    processor,
    formatted_prompt,
    image,
    max_tokens=2048,
    temperature=0.0,
    verbose=False,
)

print(output)
````

````python
text = output.text

# Remove markdown code fences
text = text.replace("```json", "").replace("```", "").strip()

data = json.loads(text)
````
